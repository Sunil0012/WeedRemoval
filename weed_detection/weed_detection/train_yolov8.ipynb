```python
import yaml

train_images_path = './dataset/images/train'
val_images_path = './dataset/images/val'
test_images_path = './dataset/images/test'
classes = ['Crop', 'Weed'] 

data = {
    'train': train_images_path,
    'val': val_images_path,
    'test': test_images_path,
    'nc': len(classes),
    'names': classes
}

with open('data_config.yaml', 'w') as file:
    yaml.dump(data, file, default_flow_style=False)

print("YAML file created successfully.")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt") 

model.train(data="./data_config.yaml", epochs=50) 
metrics = model.val() 
path = model.export(format="onnx") 
```

```python
from ultralytics import YOLO

def main():
    model_path = "./yolov8n.pt"
    data_yaml = "./data_config.yaml"

    model = YOLO(model_path)

    metrics = model.val(
        data=data_yaml,
        split="test",   # "val" or "test"
        imgsz=640,
        device=0,
        conf=0.25,
        iou=0.6
    )

    # Main summary (common ones)
    print("mAP50-95:", metrics.box.map)     # mean AP @ 0.50:0.95
    print("mAP50   :", metrics.box.map50)
    print("mAP75   :", metrics.box.map75)
    print("Precision:", metrics.box.mp)
    print("Recall   :", metrics.box.mr)

    # Per-class table (Ultralytics provides per-class summaries)
    # This is supported via the underlying metrics utilities. :contentReference[oaicite:7]{index=7}
    per_class = metrics.box.class_result
    # Some versions return dict-like; safest: print and inspect structure
    print("Per-class raw:", per_class)

    # If you want a clean CSV from results.csv (epoch-wise training log):
    # results.csv lives in your run folder; adjust path
    # df = pd.read_csv("runs/detect/weed_yolo11n/results.csv")
    # df.to_csv("runs/detect/weed_yolo11n/results_clean.csv", index=False)

if __name__ == "__main__":
    main()
```
