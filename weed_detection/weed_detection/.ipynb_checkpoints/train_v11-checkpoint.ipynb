```python
import os
import cv2
import yaml
import matplotlib.pyplot as plt
from collections import Counter
from ultralytics import YOLO
```

```python
# Read classes.txt
with open('/kaggle/input/weeds-detection/dataset/classes.txt', 'r', encoding='utf-8') as f:
    classes = [line.strip() for line in f.readlines() if line.strip()]

print(f"Number of classes: {len(classes)}")
print(f"Classes: {classes}")  # Example output: ['crop', 'weed'] or more
```

```python
# Count the number of images in each split
def count_images(folder):
    base_path = '/kaggle/input/weeds-detection/dataset/images'
    return len(os.listdir(os.path.join(base_path, folder)))

print(f"Number of training images: {count_images('train')}")
print(f"Number of validation images: {count_images('val')}")
print(f"Number of test images: {count_images('test')}")
```

```python
# Class distribution in label files (read each label .txt)
def count_classes_in_labels(split, classes, base_path='/kaggle/input/weeds-detection/dataset/labels'):
    label_dir = os.path.join(base_path, split)
    class_counts = Counter()

    for txt_file in os.listdir(label_dir):
        if not txt_file.endswith('.txt'):
            continue
        with open(os.path.join(label_dir, txt_file), 'r', encoding='utf-8') as f:
            for line in f:
                parts = line.strip().split()
                if not parts:
                    continue
                class_id = int(parts[0])
                class_counts[class_id] += 1

    # Map class IDs to class names safely
    return {classes[k] if k < len(classes) else str(k): v for k, v in class_counts.items()}

print("Train class distribution:", count_classes_in_labels('train', classes))
print("Validation class distribution:", count_classes_in_labels('val', classes))
print("Test class distribution:", count_classes_in_labels('test', classes))
```

```python
# Show an example image
img_path = '/kaggle/input/weeds-detection/dataset/images/test/agri_0_1020.jpeg'
img = cv2.imread(img_path)

if img is None:
    raise FileNotFoundError(f"Image not found or cannot be read: {img_path}")

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.title("Example Image")
plt.axis("off")
plt.show()
```

```python
from ultralytics import YOLO

# Load a COCO-pretrained YOLO11n model
model = YOLO("yolo11n.pt")
```

```python
# Example test image
test_img = '/kaggle/input/weeds-detection/dataset/images/test/agri_0_1130.jpeg'

# Run inference
results = model(source=test_img, save=True, conf=0.3)[0]

# Show results (image with bounding boxes)
results.show()

print(f"Detected {len(results.boxes)} objects")
for box in results.boxes:
    cls_name = classes[int(box.cls)]
    conf = float(box.conf)
    print(f"{cls_name} - Confidence: {conf:.2f}")
```

```python
# Prepare YAML content (YOLO standard format)
data_root = "/kaggle/input/weeds-detection/dataset"

data_yaml = {
    "path": os.path.abspath(data_root),  # Absolute path is safer
    "train": "images/train",
    "val": "images/val",
    "test": "images/test",
    "names": {i: name for i, name in enumerate(classes)},
    # "nc": len(classes)  # optional (Ultralytics can infer this from names)
}

# Save YAML file
yaml_path = "/kaggle/working/weeds.yaml"  # change to your desired location
with open(yaml_path, "w", encoding="utf-8") as f:
    yaml.dump(data_yaml, f, default_flow_style=False, sort_keys=False)

print(f"YAML file saved successfully: {yaml_path}")
```

```python
# Display the contents of the YAML file
with open(yaml_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# Load pretrained YOLO11 nano model
model = YOLO("yolo11n.pt")  # latest pretrained nano

# Start training (parameters tuned)
results = model.train(
    data="/kaggle/working/weeds.yaml",  # the YAML file you saved
    epochs=100,
    patience=20,            # early stopping
    imgsz=640,
    batch=16,               # increase if GPU memory allows
    augment=True,
    degrees=10.0,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.1,
    copy_paste=0.1,
    conf=0.25,
    iou=0.6,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,
    warmup_epochs=3,
    close_mosaic=10,
    amp=True,               # mixed precision for faster training
    project="weed_detection_runs",
    name="weed_detection_yolo11n_custom",
    exist_ok=True,
    plots=True,
    save=True,
    verbose=True
)

print("Training completed!")
print("Best model saved at: weed_detection_runs/detect/weed_detection_yolo11n_custom/weights/best.pt")
```

```python
# Load the best model (write the correct path after training)
best_model_path = "/kaggle/working/weed_detection_runs/weed_detection_yolo11n_custom/weights/best.pt"
model = YOLO(best_model_path)

# Get class names (automatically loaded from the trained model)
classes = model.names
print("Classes:", classes)
```

```python
val_metrics = model.val(
    data="/kaggle/working/weeds.yaml",   # your YAML file
    split="val",                         # validation split
    imgsz=640,
    batch=16,
    conf=0.001,                          # low conf to collect all detections (good for PR curve)
    iou=0.6,
    plots=True,                          # generates PR curves + confusion matrix
    save_json=False,                     # set True to save COCO-style JSON
    project="weed_detection_runs",
    name="val_evaluation",
    exist_ok=True,
    verbose=True                         # detailed per-class output
)

print("\n=== OVERALL VALIDATION METRICS ===")
print(f"mAP@0.5:0.95 : {val_metrics.box.map:.4f}")
print(f"mAP@0.5      : {val_metrics.box.map50:.4f}")
print(f"mAP@0.75     : {val_metrics.box.map75:.4f}")
print(f"Mean Precision (mP): {val_metrics.box.mp:.4f}")
print(f"Mean Recall    (mR): {val_metrics.box.mr:.4f}")

# Per-class metrics
print("\n=== PER-CLASS METRICS (Validation) ===")
for i, class_name in classes.items():
    print(
        f"{class_name}: "
        f"Precision={val_metrics.box.p[i]:.4f}, "
        f"Recall={val_metrics.box.r[i]:.4f}, "
        f"mAP@0.5:0.95={val_metrics.box.maps[i]:.4f}"
    )
```

```python
test_metrics = model.val(
    data="/kaggle/working/weeds.yaml",
    split="test",                        # test split
    imgsz=640,
    batch=16,
    conf=0.001,
    iou=0.6,
    plots=True,                          # saves all plots (to a separate folder)
    save_json=False,
    project="weed_detection_runs",
    name="test_evaluation",
    exist_ok=True,
    verbose=True
)

print("\n=== OVERALL TEST METRICS ===")
print(f"mAP@0.5:0.95 : {test_metrics.box.map:.4f}")
print(f"mAP@0.5      : {test_metrics.box.map50:.4f}")
print(f"mAP@0.75     : {test_metrics.box.map75:.4f}")
print(f"Mean Precision (mP): {test_metrics.box.mp:.4f}")
print(f"Mean Recall    (mR): {test_metrics.box.mr:.4f}")

# Per-class metrics
print("\n=== PER-CLASS METRICS (Test) ===")
for i, class_name in classes.items():
    print(
        f"{class_name}: "
        f"Precision={test_metrics.box.p[i]:.4f}, "
        f"Recall={test_metrics.box.r[i]:.4f}, "
        f"mAP@0.5:0.95={test_metrics.box.maps[i]:.4f}"
    )
```

```python
# Function: Display all .png and .jpg files in a folder
def show_plots(folder_path):
    if not os.path.exists(folder_path):
        print(f"Folder not found: {folder_path}")
        return

    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith((".png", ".jpg", ".jpeg"))]
    image_files.sort()  # show in order

    print(f"\nImages in folder: {folder_path}")
    for img_file in image_files:
        img_path = os.path.join(folder_path, img_file)
        img = plt.imread(img_path)

        plt.figure(figsize=(10, 8))
        plt.imshow(img)
        plt.title(img_file)
        plt.axis("off")
        plt.show()
```

```python
# Validation plots
val_plot_dir = "weed_detection_runs/val_evaluation"
show_plots(val_plot_dir)
```

```python
# Test plots
test_plot_dir = 'weed_detection_runs/test_evaluation'
show_plots(test_plot_dir)
```

```python
def show_all_plots(folder):
    if os.path.exists(folder):
        files = sorted([f for f in os.listdir(folder) if f.endswith(('.png', '.jpg'))])
        for file in files:
            path = os.path.join(folder, file)
            img = plt.imread(path)
            plt.figure(figsize=(12, 8))
            plt.imshow(img)
            plt.title(file)
            plt.axis('off')
            plt.show()

show_all_plots('/kaggle/working/weed_detection_runs/val_evaluation')
show_all_plots('/kaggle/working/weed_detection_runs/test_evaluation')
```

```python
import os
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO

# 1) Load the best model (double-check the path!)
best_model_path = "/kaggle/working/weed_detection_runs/weed_detection_yolo11n_custom/weights/best.pt"

if not os.path.exists(best_model_path):
    print(f"⚠️ Model not found: {best_model_path}")
    print("List training folders to locate best.pt:")
    !ls /kaggle/working/weed_detection_runs/ -R | grep best.pt
else:
    print(f"✅ Model found: {best_model_path}")

model = YOLO(best_model_path)

# 2) Test image
image_path = "/kaggle/input/weeds-detection/dataset/images/test/agri_0_1020.jpeg"

if not os.path.exists(image_path):
    print(f"⚠️ Image not found: {image_path}")
    print("First 20 images in test folder:")
    !ls /kaggle/input/weeds-detection/dataset/images/test/ | head -20
else:
    print(f"✅ Image found: {os.path.basename(image_path)}")

# 3) Prediction
results = model.predict(
    source=image_path,
    conf=0.25,      # lower to 0.1 if you want more detections
    iou=0.45,
    imgsz=640,
    save=False,
    show=False
)[0]

# 4) Print detections + class names
print(f"\n🔍 Number of detected objects: {len(results.boxes)}")
print(f"Classes: {results.names}")  # e.g. {0: 'crop', 1: 'weed'}

if len(results.boxes) > 0:
    for i, box in enumerate(results.boxes):
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        class_name = results.names[cls_id]
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        print(f" → {i+1}. {class_name} | Conf: {conf:.3f} | Box: ({x1},{y1}) - ({x2},{y2})")
else:
    print(" → No detections. Try conf=0.1 and run again.")

# 5) Visualization
img = cv2.imread(image_path)
if img is None:
    raise FileNotFoundError(f"Could not read image: {image_path}")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

for box in results.boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    cls_id = int(box.cls[0])
    conf = float(box.conf[0])
    label = f"{results.names[cls_id]} {conf:.2f}"

    cv2.rectangle(img_rgb, (x1, y1), (x2, y2), (0, 255, 0), 3)
    cv2.putText(img_rgb, label, (x1, max(0, y1 - 8)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

plt.figure(figsize=(10, 7))
plt.imshow(img_rgb)
plt.title(f"Weed Detection Prediction - {os.path.basename(image_path)}", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()
```
